```python
import pandas as pd

df = pd.read_csv("/Volumes/workspace/default/ecommerce_data/raw/e_commerce_shopper_behaviour_RAW_bronze.csv")
landing_path = "/Volumes/workspace/default/ecommerce_data/landing/"

batch_size = 15000
num_batches = len(df) // batch_size + 1

for i in range(num_batches):
    batch = df.iloc[i*batch_size : (i+1)*batch_size]
    if len(batch) == 0:
        break
    batch.to_csv(f"{landing_path}batch_{i:04d}.csv", index=False)
    print(f"[PRODUCER] Batch {i+1}/{num_batches} written: {len(batch)} rows")
```
